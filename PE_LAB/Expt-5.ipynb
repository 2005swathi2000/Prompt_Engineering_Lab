```python
sentence1 = "I hate this movie."

if "hate" in sentence1.lower():
    sentiment1 = "Negative"
else:
    sentiment1 = "Positive"

print("Zero-Shot Prompt:\n")
print(f"The sentiment of the sentence is {sentiment1}.")

sentence2 = "This phone is amazing."

if "amazing" in sentence2.lower():
    sentiment2 = "Positive"
elif "hate" in sentence2.lower():
    sentiment2 = "Negative"
else:
    sentiment2 = "Neutral"

print("\nFew-Shot Prompt:\n")
print(sentiment2)
```
