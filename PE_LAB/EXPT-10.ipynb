```python
from google import genai
from getpass import getpass

api_key = getpass("Enter your Gemini API Key: ")
client = genai.Client(api_key=api_key)

print("Gemini client initialized successfully!")
```

```python

# Experiment 3: Chain-of-Thought & Task Decomposition

from google import genai

# Use the client created in Step 4
MODEL = "gemini-2.5-flash"

problem = (
    "A train travels 60 km in 1 hour. "
    "How far will it travel in 3 hours, assuming constant speed?"
)

def ask_gemini(title, prompt):
    print("\n" + "=" * 55)
    print(title)
    print("=" * 55)

    try:
        response = client.models.generate_content(
            model=MODEL,
            contents=prompt
        )

        answer = response.text
        print(answer if answer else "No text response received.")
        return answer

    except Exception as e:
        print("Error:", type(e).__name__, "-", str(e))
        return None


# Prompt 1: Direct Answer
direct = ask_gemini(
    "PROMPT 1 - DIRECT ANSWER",
    problem + " Give only the final answer."
)


# Prompt 2: Step-by-Step Explanation
cot = ask_gemini(
    "PROMPT 2 - CHAIN-OF-THOUGHT STYLE",
    problem
    + "\nExplain the key calculation steps clearly, "
      "then give the final answer."
)


# Prompt 3: Task Decomposition
decomposition_prompt = """
Solve this problem by following the tasks in order.

Problem: A train travels 60 km in 1 hour.
How far will it travel in 3 hours, assuming constant speed?

Task 1: Identify the distance and time given.
Task 2: Calculate the speed in km/h.
Task 3: Identify the required travel time of 3 hours.
Task 4: Use distance = speed x time to calculate the answer.
Task 5: State the final answer with units.

Make sure all calculations use the required 3-hour duration.
"""

decomposed = ask_gemini(
    "PROMPT 3 - TASK DECOMPOSITION",
    decomposition_prompt
)


# Evaluation
print("\n" + "=" * 55)
print("EXPERIMENT SUMMARY")
print("=" * 55)

print("Expected answer: 180 km")
print("Direct response received:", direct is not None)
print("Step-by-step response received:", cot is not None)
print("Decomposition response received:", decomposed is not None)
```
