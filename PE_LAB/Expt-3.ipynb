```python

# Experiment: Prompt Engineering - Romeo and Juliet

from google import genai
from getpass import getpass
import time

# Create Gemini client
api_key = getpass("Enter your Gemini API Key: ")
client = genai.Client(api_key=api_key)

# Use a currently documented Gemini model
MODEL = "gemini-3.8-flash"


# Function to handle temporary server errors
def generate_response(prompt):
    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model=MODEL,
                contents=prompt
            )
            return response.text

        except Exception as e:
            error = str(e)

            if ("503" in error or "429" in error) and attempt < 2:
                wait = 5 * (attempt + 1)
                print(f"Temporary API error. Retrying in {wait} seconds...")
                time.sleep(wait)
            else:
                print("API Error:", error)
                return None


# Prompt 1 - Minimal Instruction
print("Prompt 1 - Minimal Instruction:\n")

response1 = generate_response(
    "Summarize Romeo and Juliet."
)

if response1:
    print(response1)


# Prompt 2 - More Specific Instruction
print("\nPrompt 2 - More Specific Instruction:\n")

response2 = generate_response(
    "Summarize Romeo and Juliet in five sentences."
)

if response2:
    print(response2)


# Prompt 3 - Refined Instruction
print("\nPrompt 3 - Refined Instruction:\n")

response3 = generate_response("""
Summarize Romeo and Juliet in two sim|ple sentences.
Use easy language suitable for school students.
""")

if response3:
    print(response3)


# Experiment completed
print("\nExperiment execution finished.")
```
